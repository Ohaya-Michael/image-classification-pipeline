```python
import tensorflow as tf
import numpy as np
import os

# Define your custom logic
def custom_loss(y_true, y_pred):
    return tf.reduce_mean(tf.square(y_true - y_pred))
```

```python
def evaluate_model(test_image, model_path):
    '''
    Receives
    -> testDir: A list of testing images
    -> model_path: The path to the trained model
    -> numImages: The number of images to be used for prediction

    Returns null
    Displays an image with its predicted class and the actual class for the test dataset
    '''
    print("Evaluating the model ...")
    top_5 = {}

    # Load the model and provide the mapping
    model = tf.keras.models.load_model(
        model_path, 
        custom_objects={'custom_loss': custom_loss}
    )

    # Optional: Define class names if you have them
    class_names = ['daisy', 'dandelion', 'roses', 'sunflowers', 'tulips']

    # 1. Bild laden (Größe an dein Modell anpassen!)
    img_path = os.getcwd() + "/" + test_image
    target_size = (240, 240) # Beispielgröße
    img = tf.keras.utils.load_img(img_path, target_size=target_size)

    # 2. In Array umwandeln
    img_array = tf.keras.utils.img_to_array(img)

    # 4. Batch-Dimension hinzufügen (von (240, 240, 3) zu (1, 240, 240, 3))
    img_array = np.expand_dims(img_array, axis=0)

    # Vorhersage ausführen
    predictions = model.predict(img_array)[0] # [0], um aus der Batch-Dimension (1, N) zu (N,) zu kommen

    # Die Top-5 Indizes finden (sortiert nach Wahrscheinlichkeit)
    top_5_indices = np.argsort(predictions)[-5:][::-1]


    for i in top_5_indices:
        score = predictions[i]
        # Falls class_names existiert: label = class_names[i]
        # Sonst nutzen wir den Index:
        label = class_names[i] if i < len(class_names) else f"Class {i}"
        top_5[label] = f"{score:.2%}"
        
    return top_5
```

```python
evaluate_model("10791227_7168491604.jpg", "buildEfficientB0_Model/trained_model_finetuned.keras")
```
